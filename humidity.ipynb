```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import linear_model
import seaborn as sns
```

```python
df = pd.read_excel('/content/drive/MyDrive/Linear regreesion/1.xlsx')
print(df)
```

```python
plt.scatter(df.Humidity,df.Rainfall)
plt.xlabel('Humidity')
plt.ylabel('Rainfall')
plt.show()
```

```python
pkr=linear_model.LinearRegression()
pkr.fit(df[['Humidity']],df.Rainfall)
```

```python
pkr.score(df[['Humidity']],df.Rainfall)
```

```python
pkr.predict([[43]])
```

```python
plt.scatter(df.Humidity,df.Rainfall)
plt.xlabel('Humidity')
plt.ylabel('Rainfall')
plt.plot(df.Humidity,pkr.predict(df[['Humidity']]),color='red')
plt.show()
```

```python
db= pd.read_excel('/content/drive/MyDrive/Linear regreesion/diabetics.xlsx')
db["Weight"] = db["Weight"].fillna(db["Weight"].mean())
db["Height"] = db["Height"].fillna(db["Height"].mean())
db = db.dropna(subset=["Diabetics"])
pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',None)
pd.set_option('display.width',None)
print(db)
```

```python
plt.scatter(db["Age"], db["Weight"])
plt.xlabel("Age")
plt.ylabel("Weight")
plt.show()

plt.scatter(db["Age"], db["Height"])
plt.xlabel("Age")
plt.ylabel("Height")
plt.show()

plt.scatter(db["Age"], db["BMI"])
plt.xlabel("Age")
plt.ylabel("BMI")
plt.show()

plt.scatter(db["Age"], db["Diabetics"])
plt.xlabel("Age")
plt.ylabel("Diabetics")
plt.show()
```

```python
lr=linear_model.LinearRegression()
lr.fit(db[['Age','Weight','Height','BMI']],db.Diabetics)
```

```python
lr.score(db[['Age','Weight','Height','BMI']],db.Diabetics)
```

```python
plt.scatter(db["Age"], db["Weight"])
plt.xlabel("Age")
plt.ylabel("Weight")
plt.plot(db.Age,lr.predict(db[['Age','Weight','Height','BMI']]),color='red')
plt.show()

plt.scatter(db["Age"], db["Height"])
plt.xlabel("Age")
plt.ylabel("Height")
plt.show()

plt.scatter(db["Age"], db["BMI"])
plt.xlabel("Age")
plt.ylabel("BMI")
plt.show()

plt.scatter(db["Age"], db["Diabetics"])
plt.xlabel("Age")
plt.ylabel("Diabetics")
plt.show()
```

```python
lr.predict([[ 28 , 112 , 4 ,67.1 ]])
```
