```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
```

```python
df = pd.read_excel('/content/drive/MyDrive/Linear regreesion/house_price_regression_dataset.xlsx')
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', None)
print(df)
```

```python
x= df[['Area_sqft','Bedrooms','Age_years','Distance_to_CityCenter_km']]
y= df[['Price_Lakh_PKR']]
```

```python
from sklearn import model_selection
x_train , x_test , y_train , y_test = model_selection.train_test_split(x,y ,test_size=0.2)
```

```python
len(x_test)
```

```python
lr= linear_model.LinearRegression()
lr.fit(x_train,y_train)
```

```python


lr.predict(x_test)
```

```python
lr.score(x_test,y_test)
```

```python
plt.scatter(x_test['Area_sqft'],y_test)
plt.scatter(x_test['Bedrooms'],y_test)
plt.scatter(x_test['Age_years'],y_test)
plt.scatter(x_test['Distance_to_CityCenter_km'],y_test)
plt.legend()
#plt.plot(x_test,lr.predict(x_test))
plt.show()
```
