```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn import linear_model
```

```python
df = pd.read_excel('/content/drive/MyDrive/Linear regreesion/humidity_rain_regression (1).xlsx')
print(df)
```

```python
plt.scatter(df.Humidity,df.Rainfall, color="red")
plt.xlabel("Humidity")
plt.ylabel("Rainfall")
plt.show()
```

```python
lr=linear_model.LinearRegression()
lr.fit(df[['Humidity']],df.Rainfall)
```

```python
lr.score(df[['Humidity']],df.Rainfall)
```

```python
plt.scatter(df.Humidity,df.Rainfall, color="red")
plt.xlabel("Humidity")
plt.ylabel("Rainfall")
plt.plot(df.Humidity,lr.predict(df[['Humidity']]),color="blue")
plt.show()
```
